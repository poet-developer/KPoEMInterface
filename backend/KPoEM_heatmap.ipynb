## KPoEM Emotion Encoding

```python
# 44 Emotions categories from original KOTE dataset
emotions = [
    '불평/불만','환영/호의','감동/감탄','지긋지긋','고마움','슬픔','화남/분노','존경','기대감',
    '우쭐댐/무시함','안타까움/실망','비장함','의심/불신','뿌듯함','편안/쾌적','신기함/관심','아껴주는',
    '부끄러움','공포/무서움','절망','한심함','역겨움/징그러움','짜증','어이없음','없음','패배/자기혐오',
    '귀찮음','힘듦/지침','즐거움/신남','깨달음','죄책감','증오/혐오','흐뭇함(귀여움/예쁨)','당황/난처',
    '경악','부담/안_내킴','서러움','재미없음','불쌍함/연민','놀람','행복','불안/걱정','기쁨','안심/신뢰'
]

# emotions1 = ['dissatisfaction','welcome','admiration','fed_up','gratitude','sadness','anger','respect',
# 'expectancy','arrogance','disappointment','resolute','distrust','pride','comfort','interest','care',
# 'shame','fear','despair','pathetic','disgust','irritation','preposterous','no_emotion','self_loathing','laziness',
# 'exhaustion','excitement','realization','guilt','contempt','adoration','embarrassment','shock','reluctant',
# 'sorrow','boredom','compassion','surprise','happiness','anxiety','joy','relief'
# ] # 영어 라벨
```

```python
annot_cols = [f"annotator_0{i}" for i in range(1, 6)]
```

```python
poet = "김소월"
```

```python
# 1. 특정 시인 작품만 필터링
df_poet = df_all[df_all["poet"] == poet].copy()
```

```python
# 2. 감정 → 벡터 함수
emotion_to_idx = {e: i for i, e in enumerate(emotions)}

def emotions_to_vector(emotion_string):
    vec = np.zeros(len(emotions))
    if pd.isna(emotion_string):
        return vec
    
    for item in emotion_string.split(","):
        item = item.strip()
        if item in emotion_to_idx:
            vec[emotion_to_idx[item]] = 1
    return vec
```

```python
# 3. annotator별 원핫 벡터 생성 - annotator 판단을 수치화하는 단계
for a in annot_cols:
    df_poet[a + "_vec"] = df_poet[a].apply(emotions_to_vector)
```

```python
# 4. 작품별 감정 벡터 평균 계산
poem_vectors = []

for poem_id, group in df_poet.groupby("poem_id"):
    annot_avgs = [] # annotator별 벡터 평균 저장 리스트
    for a in annot_cols:
        avg = np.stack(group[a + "_vec"].values).mean(axis=0) # 데이터 별 평균 (shape: 44,)
        annot_avgs.append(avg)
    
    poem_vector = np.stack(annot_avgs).mean(axis=0) # annotator 평균 → 최종 작품 벡터
    poem_vectors.append([poem_id] + poem_vector.tolist())
```

```python
# 5. DataFrame으로 변환
poem_df_poet = pd.DataFrame(poem_vectors, columns=["poem_id"] + emotions)

poem_df_poet.head()
```

```python
title_map = df_poet[["poem_id", "title"]].drop_duplicates()
```

```python
poem_df = poem_df_poet.merge(
    title_map,
    on="poem_id",
    how="left"
)
```

## KPoEM Emotion Heatmap Construction

```python
import numpy as np

# 작품 정렬(중요: 슬라이드가 예쁘게 넘어가려면 순서 고정)
subset = subset.sort_values("poem_id").reset_index(drop=True)

titles = title_map["title"].astype(str).tolist()          # x축 라벨
M = subset[emotions].to_numpy().T                      # (44 emotions, 165 poems)
n_poems = M.shape[1]
n_emotions = M.shape[0]
```

```python
!pip install -U plotly nbformat
```

```python
import plotly.graph_objects as go
import numpy as np

WINDOW = 50
frames = []

for start in range(0, n_poems - WINDOW + 1):
    end = start + WINDOW
    frames.append(
        go.Frame(
            data=[
                go.Heatmap(
                    z=M[:, start:end],
                    x=titles[start:end],
                    y=emotions,
                    colorscale="Blues",
                    zmin=0,
                    zmax=1
                )
            ],
            name=str(start)
        )
    )

fig = go.Figure(
    data=[
        go.Heatmap(
            z=M[:, :WINDOW],
            x=titles[:WINDOW],
            y=emotions,
            colorscale="Blues",
            zmin=0,
            zmax=1
        )
    ],
    frames=frames
)
```

```python
fig.update_layout(
    title="KPoEM – 김소월 작품단위 감정 히트맵",
    width=1400,
    height=1100,
    autosize=False,
    margin=dict(l=100, r=100, t=80, b=250),

    xaxis=dict(
        tickangle=60,
        fixedrange=True,
        automargin=False
    ),
    yaxis=dict(
        autorange="reversed",
        tickmode="array",
        tickvals=emotions,
        ticktext=emotions,
        fixedrange=True,
        automargin=False,
        domain=[0.0, 0.9]
    ),

    uirevision="fixed",

    sliders=[{
        "currentvalue": {"visible": False},
        "steps": [
            {
                "method": "animate",
                "args": [
                    [str(i)],
                    {
                        "mode": "immediate",
                        "frame": {"duration": 0, "redraw": True},
                        "transition": {"duration": 0}
                    }
                ],
                "label": ""
            }
            for i in range(0, n_poems - WINDOW + 1)
        ]
    }]
)
```

```python
# HTML 저장
fig.write_html("kpoem_heatmap_slider_kim.html")
```
